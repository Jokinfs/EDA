1.1.- LIMPIEZA DATOS BITCOIN


Fuente: https://es.investing.com/crypto/bitcoin

```python
# Importamos el fichero de los datos historicos del bitcoin y creamos un dataframe con formato de miles con ".", decimales con "," y fecha como str.

df_btc = pd.read_csv(
    "../data/Datos históricos del Bitcoin.csv",
    thousands=".",
    decimal=",",
    dtype={"Fecha": str}
)
```

1.4.- LIMPIEZA DATOS FED

Fuente: https://fred.stlouisfed.org/series/fedfunds

```python
df_fed = pd.read_csv("../data/FEDFUNDS.csv", dtype={"observation_date": str})
df_fed = df_fed.rename(columns={"observation_date": "Fecha"})
df_fed.info()
df_fed
```

1.5.- TABLAS CON LAS MISMAS FECHAS DE DATOS

```python
# Fechas comunes: solo los días que existen en BTC, S&P 500 y oro, a partir del 01-01-2014.
# El Bitcoin cotiza todos los dias pero el SP500 y el Oro no, cierra fines de semana y festivos.
# La FED es mensual (día 1 de cada mes): su tipo se mantiene todo el mes, así que se rellena hacia delante (ffill) en esas fechas.

df_fed["Fecha"] = pd.to_datetime(df_fed["Fecha"], format="%Y-%m-%d")
df_fed = df_fed.set_index("Fecha").sort_index()

fechas_comunes = df_btc.index.intersection(df_sp500.index).intersection(df_oro.index)
fechas_comunes = fechas_comunes[fechas_comunes >= "2014-01-01"]

df_btc = df_btc.loc[fechas_comunes]
df_sp500 = df_sp500.loc[fechas_comunes]
df_oro = df_oro.loc[fechas_comunes]
df_fed = df_fed.reindex(df_fed.index.union(fechas_comunes)).ffill().loc[fechas_comunes]

```

```python
for nombre, df in [("BTC", df_btc), ("S&P 500", df_sp500), ("Oro", df_oro), ("FED", df_fed)]:
    print(nombre, df.shape, df.index.min().date(), "->", df.index.max().date())
```

***1.- ¿Cómo se comportó el precio del bitcoin en las crisis financieras ocurridas desde 2014?***

```python
# Hacemos copia de la tabla limpia de bitcoin e incluimos con una columna que indicará la crisis de cada día.

df_btc_h1 = df_btc.copy()
df_btc_h1["crisis"] = "Sin crisis"
df_btc_h1
```

```python
# Se recorre la tabla de crisis y se marcan los días que caen entre su inicio y su fin.

for i in range(len(crisis)):
    inicio = crisis["inicio"][i]
    fin = crisis["fin"][i]
    nombre = crisis["crisis"][i]
    df_btc_h1.loc[inicio:fin, "crisis"] = nombre
```

```python
df_btc_h1["crisis"].value_counts()
```

```python
# Precio de cierre del primer y del último día de cada crisis, y variación entre ambos.

tabla_btc = df_btc_h1[df_btc_h1["crisis"] != "Sin crisis"].groupby("crisis").agg(
    precio_inicio=("Último", "first"),
    precio_fin=("Último", "last")
)
tabla_btc["variacion_%"] = (tabla_btc["precio_fin"] - tabla_btc["precio_inicio"]) / tabla_btc["precio_inicio"] * 100
tabla_btc = tabla_btc.reindex(crisis["crisis"])
tabla_btc
```

Grafico con Seaborn

```python
import seaborn as sns
import matplotlib.pyplot as plt
```

```python
sns.set_theme(style="whitegrid")
```

```python
# Figura 1.1: evolución del precio del bitcoin (en miles de USD) con los periodos de crisis marcados.

btc_miles = df_btc["Último"] / 1000

plt.figure(figsize=(16, 8))

# Línea completa del precio
sns.lineplot(x=btc_miles.index, y=btc_miles, color="steelblue", label="Bitcoin")

for i in range(len(crisis)):
    inicio = crisis["inicio"][i]
    fin = crisis["fin"][i]
    nombre = crisis["crisis"][i]

    # Franja sombreada del periodo de crisis
    plt.axvspan(inicio, fin, color="red", alpha=0.15)

    # Tramo del precio durante la crisis, en rojo
    tramo = btc_miles.loc[inicio:fin]
    sns.lineplot(x=tramo.index, y=tramo, color="red", linewidth=2.5)

    # Nombre de la crisis, en vertical en la parte superior
    plt.text(inicio, 350, nombre, rotation=90, va="top", ha="right", color="darkred", fontsize=10)

plt.yscale("log")
plt.ylim(0.15, 400)
plt.yticks([0.2, 0.5, 1, 2, 5, 10, 20, 50, 100],
           ["0,2", "0,5", "1", "2", "5", "10", "20", "50", "100"])

plt.title("Figura 1.1 - Evolución del precio del bitcoin y crisis financieras (2014-2026)")
plt.xlabel("Fecha")
plt.ylabel("Precio (miles de USD, escala logarítmica)")
plt.show()
```

4.- Correlaciones Bitcoin-Oro-SP500

```python
# Variación diaria (%) recalculada sobre las fechas comunes, para que los tres activos
# comparen siempre el mismo par de días (la de Investing del bitcoin incluye fines de semana).

var_diaria = pd.DataFrame(index=df_btc.index[1:])

for nombre, tabla in [("bitcoin", df_btc), ("sp500", df_sp500), ("oro", df_oro)]:
    cierre = tabla["Último"].values
    var_diaria[nombre] = (cierre[1:] - cierre[:-1]) / cierre[:-1] * 100

var_diaria.head()
```

```python
# Figura 2.8: matriz de correlación de las variaciones diarias de los tres activos.

correlaciones = var_diaria.rename(columns={"bitcoin": "Bitcoin", "sp500": "S&P 500", "oro": "Oro"}).corr()

f, ax = plt.subplots(figsize=(9, 7))
sns.heatmap(correlaciones,
            annot=True,
            fmt=".2f",
            linewidths=.5,
            ax=ax,
            vmin=-1,
            vmax=1,
            cmap="coolwarm")

ax.set_title("Figura 2.8 - Correlación de las variaciones diarias (2014-2026)")
plt.show()
```

Se observa que la correlación diaria del bitcoin con el oro es de 0,04 un valor muy escaso si lo comparamos con la correlación del bitcoin y el SP500 que es de 0,12.

```python
# Unión de la tabla del bitcoin (precio de cierre) y la de la FED por fecha.
# Las dos tablas ya tienen el mismo índice (fechas comunes), así que se concatenan por columnas.

df_btc_fed = pd.concat([df_btc["Último"], df_fed["FEDFUNDS"]], axis=1)
df_btc_fed = df_btc_fed.rename(columns={"Último": "bitcoin", "FEDFUNDS": "tipo_fed"})

# Fase de la política monetaria: se compara el tipo con el de hace ~3 meses (63 sesiones).
# Si ha subido más de 0,15 puntos -> "Subida"; si ha bajado más de 0,15 -> "Bajada"; si no -> "Estable".
cambio_3m = df_btc_fed["tipo_fed"].diff(63)
df_btc_fed["fase_fed"] = "Estable"
df_btc_fed.loc[cambio_3m > 0.15, "fase_fed"] = "Subida"
df_btc_fed.loc[cambio_3m < -0.15, "fase_fed"] = "Bajada"

print(df_btc_fed["fase_fed"].value_counts())
df_btc_fed
```

***Variación del bitcoin, el oro y el S&P 500 tras cada cambio de tipos de la FED***

```python
# Tabla resumen H3: variación (%) de cada activo en el mes siguiente a cada cambio de tipos de la FED.
# Se compara el precio del día del cambio con el de 21 sesiones después (~1 mes).
# Solo cambios de al menos 0,10 puntos (el tipo efectivo varía unas centésimas cada mes sin decisión de la FED).

sesiones = 21
precios = pd.DataFrame({
    "bitcoin": df_btc["Último"],
    "sp500": df_sp500["Último"],
    "oro": df_oro["Último"]
})
variacion_1m = (precios.shift(-sesiones) / precios - 1) * 100

cambio_tipo = df_btc_fed["tipo_fed"].diff()
fechas_cambio = cambio_tipo[cambio_tipo.abs() >= 0.1].index

resumen_h3 = pd.DataFrame({
    "tipo_antes": (df_btc_fed["tipo_fed"] - cambio_tipo).loc[fechas_cambio],
    "tipo_despues": df_btc_fed["tipo_fed"].loc[fechas_cambio],
    "movimiento": np.where(cambio_tipo.loc[fechas_cambio] > 0, "Subida", "Bajada")
})
resumen_h3 = resumen_h3.join(variacion_1m.loc[fechas_cambio]).dropna().round(2)
resumen_h3.index.name = "fecha_cambio"
resumen_h3
```

```python
# Variación media (%) de cada activo en el mes siguiente a una subida y a una bajada de tipos.

resumen_h3.groupby("movimiento")[["bitcoin", "sp500", "oro"]].mean().round(2)
```

```python
# Tabla resumen H3 agrupada por año: cuántas subidas y bajadas hubo, el tipo al empezar y al acabar,
# y la variación media (%) de cada activo en el mes siguiente a los cambios de ese año.

resumen_h3_anual = resumen_h3.groupby(resumen_h3.index.year).agg(
    subidas=("movimiento", lambda m: (m == "Subida").sum()),
    bajadas=("movimiento", lambda m: (m == "Bajada").sum()),
    tipo_inicio=("tipo_antes", "first"),
    tipo_fin=("tipo_despues", "last"),
    bitcoin=("bitcoin", "mean"),
    sp500=("sp500", "mean"),
    oro=("oro", "mean")
).round(2)
resumen_h3_anual.index.name = "año"
resumen_h3_anual
```

```python
# Figura 3.5: variación media (%) de cada activo en el mes siguiente a los cambios de tipos de cada año (tabla resumen_h3_anual).

# Etiqueta de cada año con el número de subidas y bajadas de tipos.
etiquetas_h3 = [f"{año}\n{fila.subidas} sub. / {fila.bajadas} baj." for año, fila in resumen_h3_anual.iterrows()]

# Formato largo para seaborn: una fila por año y activo.
h3_largo = (resumen_h3_anual[["bitcoin", "sp500", "oro"]]
            .rename(columns={"bitcoin": "Bitcoin", "sp500": "S&P 500", "oro": "Oro"})
            .set_axis(etiquetas_h3)
            .rename_axis("año").reset_index()
            .melt(id_vars="año", var_name="activo", value_name="variacion"))

activos_h3 = ["Bitcoin", "S&P 500", "Oro"]

plt.figure(figsize=(14, 7))
ax = sns.barplot(data=h3_largo, x="año", y="variacion", hue="activo",
                 hue_order=activos_h3,
                 palette={"Bitcoin": "steelblue", "S&P 500": "seagreen", "Oro": "goldenrod"})

# Valor (%) encima/debajo de cada barra.
for contenedor in ax.containers:
    ax.bar_label(contenedor, fmt="%.1f", padding=3, fontsize=8)

plt.axhline(0, color="black", linewidth=1)
plt.title("Figura 3.5 - Variación media de cada activo en el mes siguiente a los cambios de tipos de la FED, por año (%)")
plt.ylabel("Variación media en el mes siguiente (%)")
plt.xlabel("")
plt.legend(title="")
plt.tight_layout()
plt.show()
```

**LA HIPOTESIS 3 SE CONFIRMA PARCIALMENTE:** 
- La hipótesis 3 se confirma solo parcialmente. En los ciclos de subidas fuertes y seguidas (2018 y 2022) el bitcoin y el S&P 500 cayeron; pero de media, tras una subida el bitcoin apenas se movió (+0,4 %). Tras las bajadas de tipos subieron los tres activos, el oro incluido. Los tipos influyen en el bitcoin, pero no son lo único que mueve su precio, y su reacción es siempre más intensa que la de la bolsa.
