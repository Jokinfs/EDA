## **¿ES EL BITCOIN EL "ORO DIGITAL"? ¿ACTÚA COMO ACTIVO REFUGIO EN LAS CRISIS ECONÓMICAS?**

-------------------------------------------------------------------------------------------------------------------------------------

Un activo refugio es aquel que mantiene su valor, o incluso sube, cuando los mercados caen. El referente clásico es el oro.

En los últimos años se ha extendido la idea de que el bitcoin es el "oro digital" y en este proyecto analizaremos con datos la pregunta de:


# ***¿Se comporta el bitcoin como un verdadero activo refugio, es decir, protege el capital cuando la bolsa cae?".***




-------------------------------------------------------------------------------------------------------------------------------------

# **1.- HIPÓTESIS** 



## **HIPOTESIS GENERAL:**

*NO. EL BITCOIN NO ES UN ACTIVO REFUGIO, PORQUE EN ÉPOCAS DE CRISIS HA TENIDO MUCHA VOLATILIDAD Y SU VALOR SE HA VISTO REDUCIDO EN ESOS PERIODOS.*

2.- ¿Cómo se comportó el oro, el activo refugio por excelencia, en esos mismos momentos?

## **3.- ESTRUCTURA PARA EL ANÁLISIS DE DATOS**

El periodo de analisis de los datos comprende desde **01 de enero de 2014** a **11 de noviembre de 2026.**


**USO DE CADA DATO:**
    
- **Envolucion del precio del bitcoin.** Comportamiento y evolución del precio del bitcoin para el periodo de análisis.

- **Tipos de la Fed:** clasifican los periodos en fases de subida, bajada o tipos estables.
